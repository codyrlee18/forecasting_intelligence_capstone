```python
import duckdb
import pandas
```

```python
con = duckdb.connect(database = 'forecastintelligence.duckdb', read_only=False)
```

```python
con.execute(""" 
             CREATE TABLE IF NOT EXISTS survey (
                account_id VARCHAR,
                date DATE,
                csat_score DOUBLE,
                comment_sentiment_score DOUBLE
             );
             """)

con.execute(""" 
            COPY survey
            FROM '/Users/codyrlee/Documents/CodyLee_CapstoneProject/raw_data/survey.csv'
            (AUTO_DETECT TRUE, HEADER TRUE);
            """)
```

```python
con.execute(""" 
            CREATE TABLE IF NOT EXISTS finance (
            account_id VARCHAR,
            month DATE,
            closed_won_usd DOUBLE,
            churned_usd DOUBLE,
            actual_usd DOUBLE,
            churn BOOLEAN,
            active_customer BOOLEAN)
            """)

con.execute("""
            COPY finance
            FROM '/Users/codyrlee/Documents/CodyLee_CapstoneProject/raw_data/finance.csv'
            (AUTO_DETECT TRUE, HEADER TRUE);
            """)
```
